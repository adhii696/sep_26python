```python
# multiple inheritance
# progrm fo class Student

class Student():        # Parent Class
    def get_data(self):
        self.name = input("enter the name: ")
        self.rollno = int(input("enter the rollno: "))

    def display(self):
        print(self.name)
        print(self.rollno)

class Grade(Student):       # the Child Class inheriting from Student.
    def get_mark(self):
        self.mark = int(input("enter the mark: "))

    # Method Overriding
    def display(self):
        print(self.mark)

g = Grade()
g.get_data()
g.get_mark()
g.display()
```

```python
# constructor
class Student:
  def __init__(self, name, rollno):
    self.name = name
    self.rollno = rollno

  def display_data(self):
    print(self.name)
    print(self.rollno)

class Grade(Student):
  def __init__(self, name, rollno, mark):
    super().__init__(name, rollno)        # the super() function is used to gain access to the methods and properties of a parent (or sibling) class.
    self.mark = mark
  def display(self):
    print(self.mark)

g = Grade("abu" , 21, 65)
g.display_data()
g.display()
```

```python

###########     USING CONSTRUCTOR

class Student():
    def __init__(self, name, rollno):
        self.name = name
        self.rollno = rollno
    def display(self):
        print(self.name)
        print(self.rollno)

class Test(Student):
    def __init__(self, name, rollno, tmark):
        super().__init__(name, rollno)
        self.tmark = tmark
    def display_tmark(self):
        print(self.tmark)

class Sports():
    def __init__(self, smark):
        self.smark = smark
    def display_smark(self):
        print(self.smark)

class Grade(Test, Sports):
    def __init__(self, name, rollno, tmark, smark):
        Test.__init__(self, name, rollno, tmark)
        Sports.__init__(self, smark)
    def cal_gradeself(self):
        self.tomark = self.tmark + self.smark

        if self.tomark > 80:
            print("A grade")
        elif self.tomark > 50:
            print("B grade")
        else:
            print("You are FAILED")


z = Grade("Adhi", 101, 45, 40)

z.display()
z.display_tmark()
z.display_smark()
z.cal_gradeself()
```

```python
#operator overloading prgrm in polymorphism

# magic methods or magic function eg:   __add__ , __sub__ , __div__

# __add__ add is performed here is to do the addition
class Addition():
  def __init__(self,n):
    self.value=n
  def __add__(self,other):
    return self.value + other.value

a1=Addition(10)

a2=Addition(12)
print(a1+a2)
```
